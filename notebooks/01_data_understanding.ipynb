**Dataset:** data/raw/SUPER MARKET DATA.xlsx  
**Objective:** Load raw supermarket transaction records, inspect structural metadata, data types, missing values, duplicates, and statistical distributions.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set styling options
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.2f' % x)
plt.style.use('seaborn-v0_8-whitegrid')
```

## 1. Load Raw Dataset

```
raw_data_path = os.path.join('..', 'data', 'raw', 'SUPER MARKET DATA.xlsx')
if not os.path.exists(raw_data_path):
    raw_data_path = os.path.join('data', 'raw', 'SUPER MARKET DATA.xlsx')

df = pd.read_excel(raw_data_path)
print(f'Dataset successfully loaded with shape: {df.shape}')
```

## 2. Preview First and Last 5 Rows

```
print('--- FIRST 5 ROWS ---')
display(df.head(5))

print('--- LAST 5 ROWS ---')
display(df.tail(5))
```

## 3. Structural Metadata & Data Types

```
print('--- DATASET INFO ---')
df.info()

print('\n--- COLUMN DATA TYPES ---')
print(df.dtypes)
```

## 4. Missing Values & Duplicate Inspection

```
null_counts = df.isnull().sum()
null_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Missing_Count': null_counts, 'Missing_Percentage': null_pct})

print('--- MISSING VALUES AUDIT ---')
display(missing_df)

duplicate_count = df.duplicated().sum()
print(f'\nTotal Duplicate Records: {duplicate_count}')
```

## 5. Unique Values & Cardinality Analysis

```
cardinality_df = pd.DataFrame({
    'Data_Type': df.dtypes,
    'Unique_Values': df.nunique(),
    'Sample_Value': [df[col].iloc[0] for col in df.columns]
})

print('--- CARDINALITY SUMMARY ---')
display(cardinality_df)
```

## 6. Numerical & Categorical Descriptive Statistics

```
print('--- NUMERICAL SUMMARY ---')
display(df.describe().T)

print('\n--- CATEGORICAL SUMMARY ---')
display(df.describe(include=['O']).T)
```
