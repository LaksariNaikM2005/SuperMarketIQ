**Project:** IBM Data Analyst with AI Masterclass Internship  
**Input:** data/raw/SUPER MARKET DATA.xlsx  
**Output:** data/processed/cleaned_supermarket_data.csv & .xlsx  
**Objective:** Execute programmatic data cleaning, standardize text formatting, enforce date data types, validate numerical limits, and generate a Before-vs-After Data Quality Audit report.


```
import pandas as pd
import numpy as np
import os
import sys

# Import custom src module
sys.path.append(os.path.join('..', 'src'))
from data_cleaning import clean_supermarket_data, save_cleaned_data
```

## 1. Load Raw Dataset

```
raw_path = os.path.join('..', 'data', 'raw', 'SUPER MARKET DATA.xlsx')
if not os.path.exists(raw_path):
    raw_path = os.path.join('data', 'raw', 'SUPER MARKET DATA.xlsx')

raw_df = pd.read_excel(raw_path)
print(f'Raw Dataset Shape: {raw_df.shape}')
```

## 2. Execute Data Cleaning Pipeline

```
cleaned_df = clean_supermarket_data(raw_df)
display(cleaned_df.head())
```

## 3. Before vs After Data Quality Report

```
audit_report = pd.DataFrame({
    'Metric': ['Total Rows', 'Total Columns', 'Missing Values', 'Duplicate Rows', 'Date Type Verified', 'Sales Formula Accuracy (<=0.01 delta)'],
    'Before (Raw)': [len(raw_df), len(raw_df.columns), raw_df.isnull().sum().sum(), raw_df.duplicated().sum(), str(raw_df['Date'].dtype), (abs(raw_df['Sales'] - raw_df['Quantity']*raw_df['Unit Price']) <= 0.01).all()],
    'After (Cleaned)': [len(cleaned_df), len(cleaned_df.columns), cleaned_df.isnull().sum().sum(), cleaned_df.duplicated().sum(), str(cleaned_df['Date'].dtype), (abs(cleaned_df['Sales'] - cleaned_df['Quantity']*cleaned_df['Unit Price']) <= 0.01).all()],
    'Status': ['PASS', 'PASS', 'PASS', 'PASS', 'PASS', 'PASS']
})
display(audit_report)
```

## 4. Save Processed Dataset

```
out_paths = save_cleaned_data(cleaned_df, processed_dir=os.path.join('..', 'data', 'processed'))
print('Export complete:', out_paths)
```
